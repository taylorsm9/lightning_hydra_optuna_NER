```python
from PIL import Image


def images_to_gif(image_fnames, fname, duration):
    # image_fnames.sort(key=lambda x: int(x.name.split('_')[-2])) #sort by step
    frames = [Image.open(image) for image in image_fnames]
    frame_one = frames[0]
    frame_one.save(f'{fname}.gif', format="GIF", append_images=frames,
               save_all=True, duration=duration, loop=0)
```

```python
import os 
import re 
def list_files_in_directory(directory_path):
    """
    Lists all filenames in the given directory, sorted numerically based on a part of the filename.

    :param directory_path: Path to the directory whose files are to be listed.
    :return: A list of filenames (full paths) in the directory, sorted numerically.
    """
    if not os.path.isdir(directory_path):
        raise ValueError(f"The path '{directory_path}' is not a valid directory.")

    # List all files in the directory with their full paths
    full_paths = [os.path.join(directory_path, file) for file in os.listdir(directory_path) if os.path.isfile(os.path.join(directory_path, file))]
    
    # Sort files numerically based on a part of the filename
    # Assuming the part you want to sort by is immediately before the file extension and after the last underscore, like '..._2.png'
    sorted_paths = sorted(full_paths, key=lambda x: int(re.search(r"_(\d+)_", x).group(1)))

    return sorted_paths
```

```python
duration = 300
fname = "confusion_matrix"
data_dir = "wandb/run-20240316_103454-m1evrfbs/files/media/images"
file_list = list_files_in_directory(data_dir)
images_to_gif(file_list, fname, duration)
```
