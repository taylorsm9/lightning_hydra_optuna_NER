# Clean Log File
## Note, The log file has ANSI color codes that prevent values from displaying in a text editor, but you do not have to run this script to view the log file. You can use "less -r output.log" from the terminal. 

```python
import re

def remove_ansi_codes(file_path, output_file_path):
    # ANSI color code regex
    ansi_escape = re.compile(r'\x1B[@-_][0-?]*[ -/]*[@-~]')
    
    with open(file_path, 'r') as file:
        content = file.read()
        
    # Remove ANSI codes
    cleaned_content = ansi_escape.sub('', content)
    
    with open(output_file_path, 'w') as output_file:
        output_file.write(cleaned_content)
```

```python
file_path = '../../logs/TAC/baseline/wandb/latest-run/files/output.log'
output_file_path = '../../logs/TAC/baseline/wandb/latest-run/files/output_clean.log'
remove_ansi_codes(file_path, output_file_path)
```
