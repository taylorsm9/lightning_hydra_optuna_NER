```python
import pandas as pd
import numpy as np
from sklearn.model_selection import KFold
import os
```

```python
original_train = pd.read_pickle("../../../datasets/conll_2003/train.pkl")
original_val = pd.read_pickle("../../../datasets/conll_2003/val.pkl")

train_val_df = pd.concat([original_train, original_val])
```

```python
save_path = "../../../datasets/conll_2003/5_fold/"
```

```python
kf = KFold(n_splits=5, shuffle=True, random_state=42)
```

```python
# Dictionary to store file paths
split_files = {
    'train': [],
    'val': []
}
```

```python
for fold, (train_idx, val_idx) in enumerate(kf.split(train_val_df)):
    # Extract train and validation sets
    train_df = train_val_df.iloc[train_idx].copy()
    val_df = train_val_df.iloc[val_idx].copy()

    # Create filenames
    train_file = f"{save_path}train_fold_{fold}.pkl"
    val_file = f"{save_path}val_fold_{fold}.pkl"

    # Save pickle files using pandas
    train_df.to_pickle(train_file)
    val_df.to_pickle(val_file)

    # Store file paths
    split_files['train'].append(train_file)
    split_files['val'].append(val_file)
```

```python
for fold in range(5):
    print(f"Fold {fold}:")
    print(f"  Train: {split_files['train'][fold]}")
    print(f"  Validation: {split_files['val'][fold]}")
```
