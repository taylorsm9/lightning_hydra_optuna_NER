```python
from transformers import AutoTokenizer

# use the tokenizer of the encoder that made the predictions (model_class in the config)
tokenizer = AutoTokenizer.from_pretrained('answerdotai/ModernBERT-large')
# tokenizer = AutoTokenizer.from_pretrained('microsoft/deberta-v3-large')
# special tokens added test
# tokenizer = DebertaV2Tokenizer.from_pretrained('../../src/utilities/tokenizers/deberta_dna')
```

```python
import pandas as pd

# Load the test predictions (written by the model's test step, one row per word)
pred_file_path = '../../logs/conll/test_preds/test_data.pkl'
df = pd.read_pickle(pred_file_path)

# Inspect the first few rows to verify it's loaded correctly
print(df.head())
```

```python
# Decode individual InputIDs into tokens
# This method works with BERT, but not with Deberta (it loses the beginning of word character)
#df['Token'] = df['InputID'].apply(lambda x: tokenizer.decode([x], skip_special_tokens=False))

# Decode tokens, preserving the start-of-word marker ('▁' for DeBERTa, 'Ġ' for ModernBERT) 
df['Token'] = df['InputID'].apply(lambda x: ''.join(tokenizer.convert_ids_to_tokens(x)))
print(df)
```
